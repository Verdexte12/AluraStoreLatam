### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ingreso_total_tienda = tienda['Precio'].sum()
ingreso_total_tienda2 = tienda2['Precio'].sum()
ingreso_total_tienda3 = tienda3['Precio'].sum()
ingreso_total_tienda4 = tienda4['Precio'].sum()

print(f"Ganancia total Tienda 1: {ingreso_total_tienda}")
print(f"Ganancia total Tienda 2: {ingreso_total_tienda2}")
print(f"Ganancia total Tienda 3: {ingreso_total_tienda3}")
print(f"Ganancia total Tienda 4: {ingreso_total_tienda4}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

ganancias_tiendas = {
    'Tienda 1': ingreso_total_tienda,
    'Tienda 2': ingreso_total_tienda2,
    'Tienda 3': ingreso_total_tienda3,
    'Tienda 4': ingreso_total_tienda4
}

df_ganancias = pd.DataFrame(list(ganancias_tiendas.items()), columns=['Tienda', 'Ganancia Total'])

plt.figure(figsize=(6, 6))
ax = sns.barplot(x='Tienda', y='Ganancia Total', data=df_ganancias, palette='viridis')
plt.title('Ganancia Total por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ganancia Total')
plt.tight_layout()

for p in ax.patches:
    ax.annotate(f'{p.get_height():.2f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')

plt.show()
```

# 2. Ventas por categoría

```python
ventas_por_categoria1 = tienda['Categoría del Producto'].value_counts()
ventas_por_categoria2 = tienda2['Categoría del Producto'].value_counts()
ventas_por_categoria3 = tienda3['Categoría del Producto'].value_counts()
ventas_por_categoria4 = tienda4['Categoría del Producto'].value_counts()

print("Ventas por categoría (Tienda 1):")
display(ventas_por_categoria1)
print("\nVentas por categoría (Tienda 2):")
display(ventas_por_categoria2)
print("\nVentas por categoría (Tienda 3):")
display(ventas_por_categoria3)
print("\nVentas por categoría (Tienda 4):")
display(ventas_por_categoria4)

total_ventas_por_categoria = ventas_por_categoria1 + ventas_por_categoria2 + ventas_por_categoria3 + ventas_por_categoria4
categoria_mas_popular_total = total_ventas_por_categoria.idxmax()

print(f"\nLa categoría más popular en Tienda 1 es: {ventas_por_categoria1.idxmax()}")
print(f"La categoría más popular en Tienda 2 es: {ventas_por_categoria2.idxmax()}")
print(f"La categoría más popular en Tienda 3 es: {ventas_por_categoria3.idxmax()}")
print(f"La categoría más popular en Tienda 4 es: {ventas_por_categoria4.idxmax()}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.pie(ventas_por_categoria1, labels=ventas_por_categoria1.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribución de Categorías - Tienda 1')

plt.subplot(2, 2, 2)
plt.pie(ventas_por_categoria2, labels=ventas_por_categoria2.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribución de Categorías - Tienda 2')

plt.subplot(2, 2, 3)
plt.pie(ventas_por_categoria3, labels=ventas_por_categoria3.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribución de Categorías - Tienda 3')

plt.subplot(2, 2, 4)
plt.pie(ventas_por_categoria4, labels=ventas_por_categoria4.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribución de Categorías - Tienda 4')

plt.tight_layout()
plt.show()
```

# 3. Calificación promedio de la tienda


```python
calificacion_promedio_tienda1 = tienda['Calificación'].mean()
calificacion_promedio_tienda2 = tienda2['Calificación'].mean()
calificacion_promedio_tienda3 = tienda3['Calificación'].mean()
calificacion_promedio_tienda4 = tienda4['Calificación'].mean()

print(f"Calificación promedio Tienda 1: {calificacion_promedio_tienda1:.2f}")
print(f"Calificación promedio Tienda 2: {calificacion_promedio_tienda2:.2f}")
print(f"Calificación promedio Tienda 3: {calificacion_promedio_tienda3:.2f}")
print(f"Calificación promedio Tienda 4: {calificacion_promedio_tienda4:.2f}")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

store_ratings = {
    'Tienda 1': calificacion_promedio_tienda1,
    'Tienda 2': calificacion_promedio_tienda2,
    'Tienda 3': calificacion_promedio_tienda3,
    'Tienda 4': calificacion_promedio_tienda4
}

ratings_df = pd.DataFrame(list(store_ratings.items()), columns=['Tienda', 'Calificación Promedio'])

plt.figure(figsize=(4, 4))
ax = sns.barplot(x='Tienda', y='Calificación Promedio', data=ratings_df, palette='viridis')
plt.title('Calificación Promedio por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Calificación Promedio')
plt.ylim(0, 5)
plt.grid(axis='y')
plt.tight_layout()

for p in ax.patches:
    ax.annotate(f'{p.get_height():.2f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')

plt.show()
```

# 4. Productos más y menos vendidos

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

todas_las_tiendas = pd.concat([tienda, tienda2, tienda3, tienda4], keys=['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'], names=['Tienda', None])

recuento_productos = todas_las_tiendas.groupby('Tienda')['Producto'].value_counts().unstack(fill_value=0)

mas_vendido_por_tienda = recuento_productos.idxmax(axis=1).reset_index(name='Producto Más Vendido')
mas_vendido_por_tienda['Cantidad'] = recuento_productos.max(axis=1).values
mas_vendido_por_tienda = mas_vendido_por_tienda.rename(columns={'index': 'Tienda'}) # Renombrar la columna de índice

menos_vendido_por_tienda = recuento_productos.idxmin(axis=1).reset_index(name='Producto Menos Vendido')
menos_vendido_por_tienda['Cantidad'] = recuento_productos.min(axis=1).values
menos_vendido_por_tienda = menos_vendido_por_tienda.rename(columns={'index': 'Tienda'}) # Renombrar la columna de índice

plt.figure(figsize=(4, 4))
ax1 = sns.barplot(x='Tienda', y='Cantidad', hue='Producto Más Vendido', data=mas_vendido_por_tienda, palette='viridis', dodge=False, legend=False)
plt.title('Producto Más Vendido por Tienda con Cantidad')
plt.xlabel('Tienda')
plt.ylabel('Cantidad Vendida')
plt.tight_layout()

for index, row in mas_vendido_por_tienda.iterrows():
    ax1.text(index, row['Cantidad'], row['Producto Más Vendido'], color='black', ha="center", va="bottom", fontsize=10)
plt.show()

# Create a bar chart for the least sold products per store
plt.figure(figsize=(4, 4))
ax2 = sns.barplot(x='Tienda', y='Cantidad', hue='Producto Menos Vendido', data=menos_vendido_por_tienda, palette='viridis', dodge=False, legend=False)
plt.title('Producto Menos Vendido por Tienda con Cantidad')
plt.xlabel('Tienda')
plt.ylabel('Cantidad Vendida')
plt.tight_layout()
# Add text labels on top of each bar
for index, row in menos_vendido_por_tienda.iterrows():
    ax2.text(index, row['Cantidad'], row['Producto Menos Vendido'], color='black', ha="center", va="bottom", fontsize=10)
plt.show()
```

# 5. Envío promedio por tienda

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

costo_envio_promedio_tienda1 = tienda['Costo de envío'].mean()
costo_envio_promedio_tienda2 = tienda2['Costo de envío'].mean()
costo_envio_promedio_tienda3 = tienda3['Costo de envío'].mean()
costo_envio_promedio_tienda4 = tienda4['Costo de envío'].mean()

print(f"Costo de envío promedio Tienda 1: {costo_envio_promedio_tienda1:.2f}")
print(f"Costo de envío promedio Tienda 2: {costo_envio_promedio_tienda2:.2f}")
print(f"Costo de envío promedio Tienda 3: {costo_envio_promedio_tienda3:.2f}")
print(f"Costo de envío promedio Tienda 4: {costo_envio_promedio_tienda4:.2f}")

costos_envio_promedio = {
    'Tienda 1': costo_envio_promedio_tienda1,
    'Tienda 2': costo_envio_promedio_tienda2,
    'Tienda 3': costo_envio_promedio_tienda3,
    'Tienda 4': costo_envio_promedio_tienda4
}

df_costos_envio = pd.DataFrame(list(costos_envio_promedio.items()), columns=['Tienda', 'Costo de Envío Promedio'])

plt.figure(figsize=(4, 4))
sns.barplot(x='Tienda', y='Costo de Envío Promedio', data=df_costos_envio, palette='viridis')
plt.title('Costo de Envío Promedio por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Costo de Envío Promedio')
plt.tight_layout()
plt.show()
```
